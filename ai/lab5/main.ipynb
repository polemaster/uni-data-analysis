```python
import time
import os
import pdb
import numpy as np
import matplotlib.pyplot as plt
import dec_tree_func as dt
import sys

np.set_printoptions(threshold=sys.maxsize)
# np.random.seed(0)
# os.system('cls')

# file_name = "lenses\lenses.txt"
file_name = 'car/car.txt'
# file_name = 'tic/tic.txt'
# file_name = "letters1.txt"
# file_name = 'letters2.txt'
# file_name = 'letters3.txt'
# file_name = 'letters5.txt'

num_of_random_features = 0

number_of_experiments = 1


# part of the examples for building a tree in percentage (0-100%) other axamples can be used
# for pruning as validation subset:
part_for_constr = (
    100  # (..... you can choose the proportions for tree construction and validation )
)


print("\n\n\nbuilding tree for " + file_name)
# loading data from file and dividing it into training and test sets:
# examples with odd numbers (1,3,5,...) for training
# examples with even numbers for test

examples = dt.load_data(file_name)
# Adding worthless information:
if num_of_random_features > 0:
    examples = np.concatenate(
        (
            np.array(
                np.ceil(np.random.rand(len(examples), num_of_random_features) * 12),
                dtype=int,
            ),
            examples,
        ),
        axis=1,
    )
# print(" examples = \n" + str(examples))

[num_of_examples, num_of_columns] = examples.shape
num_of_training_examples = num_of_examples // 2
num_of_test_examples = num_of_examples - num_of_training_examples
number_for_constr = np.array(
    np.ceil(num_of_training_examples * part_for_constr / 100), dtype=int
)  # number of examples for tree construction
# print("num_of_training_examples = " + str(num_of_training_examples)+ " num_of_test_examples = "+str(num_of_test_examples))

mean_test_error = 0
mean_test_error_prun = 0


for eksp in range(number_of_experiments):

    np.random.shuffle(examples)  # random permutation of example set
    # print(" examples after shuffle = \n" + str(examples))

    train_vectors = examples[0:num_of_training_examples, 0 : num_of_columns - 1]
    train_classes = examples[0:num_of_training_examples, num_of_columns - 1]
    test_vectors = examples[
        num_of_training_examples:num_of_examples, 0 : num_of_columns - 1
    ]
    test_classes = examples[
        num_of_training_examples:num_of_examples, num_of_columns - 1
    ]

    # print("train_vectors = \n" + str(train_vectors))
    # print("train_classes = \n" + str(train_classes))
    # print("test_vectors = \n" + str(test_vectors))
    # print("test_classes = \n" + str(test_classes))

    tree = dt.build_tree(train_vectors, train_classes)

    # print("final tree = \n" + str(tree))

    dist = dt.distribution(train_vectors, train_classes, tree)
    # print("distribution = \n" + str(dist))
    d = dt.depth(tree)
    # print("depth = " + str(d))

    num_of_training_errors = dt.calc_error(train_vectors, train_classes, tree)
    num_of_test_errors = dt.calc_error(test_vectors, test_classes, tree)

    mean_test_error += num_of_test_errors / num_of_test_examples / number_of_experiments

    # pruning:
    train_vectors_constr = train_vectors[:number_for_constr]
    train_classes_constr = train_classes[:number_for_constr]
    train_vectors_valid = train_vectors[number_for_constr:]
    train_classes_valid = train_classes[number_for_constr:]

    # pruned_tree = dt.prune_tree(tree)

    # pruning:
    # a place for your algorithm here!
    # Do not use a test set for training (tree construction or pruning)!!!!!!!!!!!!!!!
    # ...............................
    # ...............................
    # ...............................
    # ...............................

    pruned_tree = tree.copy()
    [num_of_rows, num_of_nodes] = pruned_tree.shape

    num_of_training_errors_prun = dt.calc_error(
        train_vectors, train_classes, pruned_tree
    )
    num_of_test_errors_prun = dt.calc_error(test_vectors, test_classes, pruned_tree)

    print(
        "error: train = "
        + str(np.round(num_of_training_errors / num_of_training_examples, 4))
        + " test = "
        + str(np.round(num_of_test_errors / num_of_test_examples, 4))
        + "  after pruning: train = "
        + str(np.round(num_of_training_errors_prun / num_of_training_examples, 4))
        + " test = "
        + str(np.round(num_of_test_errors_prun / num_of_test_examples, 4))
    )

    mean_test_error_prun += (
        num_of_test_errors_prun / num_of_test_examples / number_of_experiments
    )

    print(tree)

print(
    "mean test error: before pruning = "
    + str(np.round(mean_test_error, 4))
    + " after pruning = "
    + str(np.round(mean_test_error_prun, 4))
    + "\nreduction of average test error = "
    + str(100 * (mean_test_error - mean_test_error_prun) / mean_test_error)
    + "%"
)
```

```python
def print_tree(tree):
    for i in range(num_of_nodes):
        toPrint = ""
        for j in range(num_of_rows):
            toPrint += str(tree[j,i]).ljust(4) + " "
        print(toPrint)

def save_tree(tree):
    with open("test.txt", 'w') as wfile:
            for i in range(num_of_nodes):
                toPrint = ""
                for j in range(num_of_rows):
                    toPrint += str(tree[j,i]).ljust(4) + " "
                wfile.write(toPrint + '\n')

def is_leaf(tree, node):
    return sum(tree[...,node]) == tree[num_of_rows-1,node]


def bfs(tree, start):
    num_of_rows, num_of_nodes = tree.shape
    num_of_rows -= 1

    stack = []
    stack.append(start)
    node_depths = np.zeros(num_of_nodes)

    while stack:
        node = stack.pop(0)
        for col in range(num_of_rows):
            if tree[col, node - 1] != 0:
                stack.append(tree[col, node - 1])
                node_depths[tree[col, node - 1]-1] = node_depths[node-1] + 1

    return node_depths

def entropy(dist_slice):
    total = sum(dist_slice)
    entropy = 0
    for i,count in enumerate(dist_slice):
        if count == 0:
            continue
        entropy -= (count/total)*np.log2(count/total)
    return entropy


def prune_by_entropy(tree, dist):
    pruned_tree = tree.copy()
    num_of_rows, num_of_nodes = tree.shape

    node_depths = bfs(tree, 1)

    to_prune = 400
    for i in range(num_of_nodes-1, -1, -1):
        if not is_leaf(pruned_tree, i):
            if node_depths[i] > 3:
                if entropy(dist[...,i]) < 1:
                    continue
                to_prune -= 1
                if to_prune == 0:
                    break

                pruned_tree[..., i] = 0
                pruned_tree[num_of_rows-1,i] = np.argmax(dist[...,i]) + 1
    return pruned_tree
```

```python
pruned_tree = prune_by_entropy(tree, dist)

[num_of_rows, num_of_nodes] = pruned_tree.shape

num_of_training_errors_prun = dt.calc_error(train_vectors,train_classes,pruned_tree)
num_of_test_errors_prun = dt.calc_error(test_vectors,test_classes,pruned_tree)

print("error: train = " + str(np.round(num_of_training_errors/num_of_training_examples,4)) + " test = " +
str(np.round(num_of_test_errors/num_of_test_examples,4)) + "  after pruning: train = " +
str(np.round(num_of_training_errors_prun/num_of_training_examples,4)) + " test = " + str(np.round(num_of_test_errors_prun/num_of_test_examples,4)))

print(f"{round((1  - num_of_test_errors_prun/num_of_test_errors)*100,2)}%")
```
