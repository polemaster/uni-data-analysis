## Data cleaning

> What problems exist in the analyzed data set?

```python
df.loc[[93,151,174,242]]
```

1. Some rows represent an aggregation of countries rather than countries (e.g. "World").
1. Some columns are redundant and can be removed (e.g. "Indicator Name").
1. For some years we do not have data for any country (e.g. 2012-2015).
1. Some countries do not have data for any year (e.g. "Taiwan", "China").

```python
# loading detailed data about countries
countries = pd.read_csv("countries_metadata.csv", encoding = "utf-8") 
```

```python
countries.head(10)
```

```python
# Combining data with country information
merge = pd.merge(df, countries, on = "Country Code")
```

```python
merge.head(10)
```

We remove rows that do not represent countries.

**Note:** The value in the *Region* column is *NaN* only when the row does not represent a country.

```python
merge = merge[pd.notnull(merge['Region'])]
```

Removing irrelevant columns.

```python
merge.columns
```

```python
merge = merge.drop(merge.columns[[60,61,64,65]], axis=1) # Note: zero indexed
merge = merge.drop('Indicator Name', axis=1)
merge = merge.drop('Indicator Code', axis=1)
```

```python
merge.columns
```

Removing columns without data.

```python
merge.count()
```

```python
merge['2015']
```

```python
merge = merge.drop(['2012', '2013', '2014', '2015'], axis=1)
```

```python
merge.count()
```

Removing countries for which there is no data.

```python
merge.mean(axis=1, numeric_only=True) 
```

```python
merge = merge[pd.notnull(merge.mean(axis=1, numeric_only=True))]
```

```python
merge.head()
```

### Task 3. (2 points)
> Which Income groups had the highest CO2 emissions per capita in 2010? What do you think, why?
>
> Draw a bar plot of average CO2 emissions by region and answer the question.

```python
# Average CO2 emissions per capita in 2010 by income group
income_avg = merge.groupby("IncomeGroup")["2010"].mean().sort_values(ascending=False)
income_avg
```

### Task 3 - answer
The highest CO2 emissions per capita in 2010 were in high income groups. This was probably caused by higher production levels in better developed countries and generally more advanced technology causes more CO2 emissions. NonOECD countries was higher than OECD countries because Middle East counties are not included in OECD organization.

```python
# Average CO2 emissions per capita in 2010 by region
region_avg = merge.groupby("Region")["2010"].mean().sort_values(ascending=False)
region_avg
```

```python
plt.figure(figsize=(10,6))
region_avg.plot(kind='bar', edgecolor='black')
plt.title("Average CO₂ Emissions per Capita by Region (2010)", fontsize=14)
plt.ylabel("CO₂ Emissions (metric tons per capita)")
plt.xlabel("Region")
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
