## **Description & Context of the dataset**

The dataset contains the first 10 minutes of approximately 10k ranked games (SOLO QUEUE) of League of Legends from a high ELO (ranking system). The ranks taken into account were DIAMOND I to MASTER. Players have roughly the same level and each game is unique.

League of Legends is a 2009 multiplayer online battle arena video game developed and published by Riot Games, in which there are 2 teams (blue and red). There are 3 lanes, a jungle, and 5 roles. The goal is to take down the enemy Nexus to win the game.

Each game is unique, and has a unique **gameId**. There are 19 features per team (38 in total) collected after 10 min in-game. The column **blueWins** is the target value, a value of 1 means the blue team has won, 0 otherwise.

Link to dataset: https://www.kaggle.com/datasets/bobbyscience/league-of-legends-diamond-ranked-games-10-min

## **Characteristics of the data set**
The dataset was collected by fetching the first 10 minutes worth of data of high rank player's ranked games. It was done via the /timeline endpoint. Getting the target variable (blueWins) was done via the /match endpoint.

 The dataset is in CSV (comma-separated values) format, it has 40 columns, and 9879 rows (not counting the first row). It consists of one set, with each individual sample having a unique gameId.

### Feature Engineering

```python
def transform(row):
  new_row = row
  new_row['diffWardsPlaced'] = new_row['blueWardsPlaced'] - new_row['redWardsPlaced']
  new_row['diffWardsDestroyed'] = new_row['blueWardsDestroyed'] - new_row['redWardsDestroyed']

  new_row['diffKills'] = new_row['blueKills'] - new_row['redKills']
  new_row['diffAssists'] = new_row['blueAssists'] - new_row['redAssists']

  new_row['blueTeamfightFactor'] = None
  if new_row['blueKills'] > 0:
    new_row['blueTeamfightFactor'] = new_row['blueAssists'] / new_row['blueKills']

  new_row['redTeamfightFactor'] = None
  if new_row['redKills'] > 0:
    new_row['redTeamfightFactor'] = new_row['redAssists'] / new_row['redKills']

  if not pd.isna(new_row['blueTeamfightFactor']) and not pd.isna(new_row['redTeamfightFactor']):
    new_row['diffTeamfight'] = new_row['blueTeamfightFactor'] - new_row['redTeamfightFactor']

  new_row['blueSpecialDeaths'] = new_row['blueDeaths'] - new_row['redKills']
  new_row['redSpecialDeaths'] = new_row['redDeaths'] - new_row['blueKills']

  new_row['diffDragons'] = new_row['blueDragons'] - new_row['redDragons']
  new_row['diffHeralds'] = new_row['blueHeralds'] - new_row['redHeralds']
  new_row['diffTowersDestroyed'] = new_row['blueTowersDestroyed'] - new_row['redTowersDestroyed']


  new_row['diffTotalGold'] = new_row['blueTotalGold'] - new_row['redTotalGold']
  new_row['diffAvgLevel'] = new_row['blueAvgLevel'] - new_row['redAvgLevel']
  new_row['diffTotalExperience'] = new_row['blueTotalExperience'] - new_row['redTotalExperience']

  new_row['diffTotalMinionsKilled'] = new_row['blueTotalMinionsKilled'] - new_row['redTotalMinionsKilled']
  new_row['diffTotalJungleMinionsKilled'] = new_row['blueTotalJungleMinionsKilled'] - new_row['redTotalJungleMinionsKilled']

  # new_row['blueWins'] = bool(new_row['blueWins'])
  # new_row['blueFirstBlood'] = bool(new_row['blueFirstBlood'])

  return new_row
```

```python
df_plus = df.apply(transform, axis=1)
```

### **Observations & Results**

#### **Blue wins correlations**:
- The highest correlation between blue wins was with difference in gold attribution - 0.51, the second-highest correlation was in number of collected experience - 0.49. Both of which could point to the importance of collecting gold and experience in the first 10 minutes of gameplay.
- Other attributes with visible positive correlation to blue’s wins (above 0.3):
  - Difference between teams in: number of killed enemies, number of assists, average level and total experience points, total minions killed. This possibly  indicates that outperforming the other team in those attributes is crucial.
  - Blue teams: kills, total gold, average level, total experience. This possibly indicates that they impact the overall game result.
- Surprisingly, the attribute that shows almost no correlation (between being high in the first 10 minutes and the end result of the game) is the difference between teams in the number of the jungle minions killed. The same is true with the overall number of the jungle minions killed by the blue team. Showing that it possibly has a very low impact on the game result.
- Interestingly, the correlation between the blue team win and the blue team first blood, while not being non-existent, is not very high (0.2), which could show that just having the first kill does not guarantee the win.
Overall, the higher correlation stems from the difference between both teams than from the blue team result itself.

#### **Blue total gold (correlations with)**:
- **blueKills**: there is a high positive correlation (0.89) between total gold of the team and its kills - It can be assumed that it results from 2 mechanisms of the game:
  - the snowballing - means that as the team gathers more gold, it is more dangerous (players acquire more items)
Award for killing a player which has a base value of 300 gold and can be increased in some special scenarios
  - blueAssists: another high correlation (0.75) between total gold is with the assists of the same team and can be assumed to result from the same mechanisms as the previous one with kills (except the reward for assist equals only 150 gold).
- **blueTowersDestroyed**: the correlation between total gold and towers destroyed by the team is weakly positive (0.35) which says that if there is some dependence between towers destroyed and gold or vice versa, it is not very strong.
- **blueAvgLevel**: the blue average level has a positive correlation of 0.62 with the blue total gold collected, which shows that in games with high value of player levels, there was also high value of gold collected.
blueTotalExperience: There seems to be a quite high positive correlation (0.68) between total gold collected in the first 10 minutes of the game and the total experience collected during that timespan.
- **blueFirstBlood**: The most significant correlation between blue’s first blood and other factors is the difference in total gold (0.38), while not being overtly high, it shows that it has an impact on the total number of gold collected in the first ten minutes.
- **blueTotalMinionsKilled**: While not being overtly high, there is a correlation of 0.28 between total gold collected and total minions killed.
blueEliteMonsters: Even though there is a possibility to collect gold from killing Dragons or Heralds, the correlation between total gold collected and number of them killed is low: 0.24

#### **Other important correlations**:
- There is a positive correlation between difference in number of kills in teams and in difference in total experience between teams (0.98), stemming from experience being related to kills in game, but not being directly translated, the same is true for difference in average level (0.77 - between diffAvgLevel and diffKills) and difference in total gold (0.92 - between diffTotalGold and diffKills). The last correlation showing the importance of kills in gold accumulation.
- The also positive correlation between difference in number of kills and difference in number of assists. This indicates that the team which has more kills will also have more assists.
- The difference in team fight factor (assists over the kills) shows positive correlation (0.56) with a difference in number of assists of teams, which is visible because in calculating team fight factor the number of assists was taken into account. The difference in team fight factor shows no significant correlation between any other attribute, meaning it does not influence any other factors.
  - Similar positive correlation is visible between the blue team's team fight factor and blue team’s number of assists (0.37).
- The difference in total number of gold shows positive correlation with difference in total experience (0.89), which shows that in games where players collected a lot of experience, they also collected a lot of gold.
- The correlation between difference in total number of gold and difference in number of assists (0.79), could show that during the games with a lot of assists, players collect a lot of gold. This factor could be influenced by the fact of assists being rewarded by gold.
- Correlation between dragons and heralds: 0.02. Such non-existent correlation is a surprising result because there could be 2 scenarios in which the correlation exists. The first scenario is where one team dominates the game and so kills both a dragon and a herald (correlation about 1). The second is where both teams are about equal, and one team kills a dragon and the second during that time kills a herald (correlation about -1). It is possible that these players are not skilled enough to do such things, or that the scenarios zero each other out.
