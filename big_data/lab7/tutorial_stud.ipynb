```python
import pyspark

sc = pyspark.SparkContext(appName="Test RDD")
# sc.master
spark = pyspark.SQLContext(sc)
```

```python
import re
txtRDD = sc.textFile("/user/adampap/tutorial")
wordsRDD = txtRDD.flatMap(lambda line: line.split(" "))
wordsRDD = txtRDD.flatMap(lambda line: re.split(r"[ \t,;\.\?!-:@\[\]\(\){}_\*/]+", line))
wordCountsRDD = wordsRDD.filter(lambda word: len(word)>0).map(lambda word: (word.lower(), 1)).reduceByKey(lambda a, b: a + b)
wordCountsRDD.sortBy(lambda pair: pair[1], ascending=False).take(30)
```

```python
df = wordCountsRDD.toDF(["word", "count"])
df.show()
```
